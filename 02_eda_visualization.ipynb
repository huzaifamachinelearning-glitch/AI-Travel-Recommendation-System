```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
# Set style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

# Load processed data
df = pd.read_csv('data/processed_dataset.csv')

print("📊 GENERATING COMPREHENSIVE EDA VISUALIZATIONS")
print("="*60)

# Create output directory
import os
os.makedirs('visualizations', exist_ok=True)

```

```python
df
```

```python
# 1. TARGET VARIABLE DISTRIBUTION
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

# Rating distribution
sns.histplot(df['rating_given'], bins=20, kde=True, ax=ax[0], color='skyblue')
ax[0].set_title('Distribution of Ratings', fontsize=14, fontweight='bold')
ax[0].set_xlabel('Rating')
ax[0].set_ylabel('Frequency')

# High vs Low rating
rating_counts = df['high_rating'].value_counts()
ax[1].pie(rating_counts, labels=['Low Rating (<4)', 'High Rating (≥4)'], 
          autopct='%1.1f%%', startangle=90, colors=['#ff9999', '#66b3ff'])
ax[1].set_title('High vs Low Rating Distribution', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig('visualizations/01_target_distribution.png', dpi=300, bbox_inches='tight')
print("✅ Saved: 01_target_distribution.png")
plt.close()

```

```python
# 2. TOP DESTINATIONS
fig, ax = plt.subplots(figsize=(12, 6))
top_places = df['place_name'].value_counts().head(10)
sns.barplot(x=top_places.values, y=top_places.index, ax=ax, palette='viridis')
ax.set_title('Top 10 Most Popular Destinations', fontsize=14, fontweight='bold')
ax.set_xlabel('Number of Visits')
ax.set_ylabel('Destination')
plt.tight_layout()
plt.savefig('visualizations/02_top_destinations.png', dpi=300, bbox_inches='tight')
print("✅ Saved: 02_top_destinations.png")
plt.close()

```

```python
# 3. PLACE TYPE ANALYSIS
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

place_type_counts = df['place_type'].value_counts()
ax[0].barh(place_type_counts.index, place_type_counts.values, color='coral')
ax[0].set_title('Distribution by Place Type', fontsize=14, fontweight='bold')
ax[0].set_xlabel('Count')

# Average rating by place type
avg_rating_by_type = df.groupby('place_type')['rating_given'].mean().sort_values()
sns.barplot(x=avg_rating_by_type.values, y=avg_rating_by_type.index, ax=ax[1], palette='coolwarm')
ax[1].set_title('Average Rating by Place Type', fontsize=14, fontweight='bold')
ax[1].set_xlabel('Average Rating')

plt.tight_layout()
plt.savefig('visualizations/03_place_type_analysis.png', dpi=300, bbox_inches='tight')
print("✅ Saved: 03_place_type_analysis.png")
plt.close()

```

```python
# 4. SEASONAL TRENDS
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

season_counts = df['season'].value_counts()
ax[0].bar(season_counts.index, season_counts.values, color=['#FFD700', '#FF6347', '#4682B4', '#32CD32'])
ax[0].set_title('Travel Distribution by Season', fontsize=14, fontweight='bold')
ax[0].set_xlabel('Season')
ax[0].set_ylabel('Number of Trips')
ax[0].tick_params(axis='x', rotation=45)

month_counts = df['month'].value_counts().reindex(
    ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
)
ax[1].plot(month_counts.index, month_counts.values, marker='o', linewidth=2, markersize=8, color='purple')
ax[1].set_title('Travel Trends by Month', fontsize=14, fontweight='bold')
ax[1].set_xlabel('Month')
ax[1].set_ylabel('Number of Trips')
ax[1].grid(True, alpha=0.3)
ax[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.savefig('visualizations/04_seasonal_trends.png', dpi=300, bbox_inches='tight')
print("✅ Saved: 04_seasonal_trends.png")
plt.close()

```

```python
# 5. BUDGET ANALYSIS
fig, ax = plt.subplots(2, 2, figsize=(14, 10))

# Budget distribution
sns.histplot(df['budget_thousand_inr'], bins=30, kde=True, ax=ax[0, 0], color='green')
ax[0, 0].set_title('Budget Distribution (in thousands INR)', fontsize=12, fontweight='bold')
ax[0, 0].set_xlabel('Budget (₹1000s)')

# Budget vs Rating
sns.scatterplot(x='budget_thousand_inr', y='rating_given', data=df, ax=ax[0, 1], alpha=0.5, color='orange')
ax[0, 1].set_title('Budget vs Rating', fontsize=12, fontweight='bold')
ax[0, 1].set_xlabel('Budget (₹1000s)')
ax[0, 1].set_ylabel('Rating')

# Trip duration distribution
sns.histplot(df['trip_duration_days'], bins=15, kde=True, ax=ax[1, 0], color='teal')
ax[1, 0].set_title('Trip Duration Distribution', fontsize=12, fontweight='bold')
ax[1, 0].set_xlabel('Duration (days)')

# Budget by companions
budget_by_companion = df.groupby('companions')['budget_thousand_inr'].mean().sort_values()
sns.barplot(x=budget_by_companion.values, y=budget_by_companion.index, ax=ax[1, 1], palette='rocket')
ax[1, 1].set_title('Average Budget by Travel Companion', fontsize=12, fontweight='bold')
ax[1, 1].set_xlabel('Average Budget (₹1000s)')

plt.tight_layout()
plt.savefig('visualizations/05_budget_analysis.png', dpi=300, bbox_inches='tight')
print("✅ Saved: 05_budget_analysis.png")
plt.close()

```

```python
# 6. CORRELATION HEATMAP
fig, ax = plt.subplots(figsize=(12, 8))
numeric_cols = df.select_dtypes(include=[np.number]).columns
correlation_matrix = df[numeric_cols].corr()
sns.heatmap(correlation_matrix, annot=False, cmap='coolwarm', center=0, 
            linewidths=0.5, ax=ax, cbar_kws={'label': 'Correlation'})
ax.set_title('Feature Correlation Heatmap', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('visualizations/06_correlation_heatmap.png', dpi=300, bbox_inches='tight')
print("✅ Saved: 06_correlation_heatmap.png")
plt.close()

```

```python
# 7. TRAVEL COMPANION & INTEREST ANALYSIS
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

companion_counts = df['companions'].value_counts()
ax[0].pie(companion_counts, labels=companion_counts.index, autopct='%1.1f%%', 
          startangle=90, colors=sns.color_palette('pastel'))
ax[0].set_title('Travel Companion Distribution', fontsize=14, fontweight='bold')

interest_counts = df['interest'].value_counts()
ax[1].barh(interest_counts.index, interest_counts.values, color='lightblue')
ax[1].set_title('Travel Interest Distribution', fontsize=14, fontweight='bold')
ax[1].set_xlabel('Count')

plt.tight_layout()
plt.savefig('visualizations/07_companion_interest.png', dpi=300, bbox_inches='tight')
print("✅ Saved: 07_companion_interest.png")
plt.close()

```

```python
# 8. ACCESSIBILITY & SAFETY
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

accessibility_counts = df['accessibility'].value_counts()
ax[0].bar(accessibility_counts.index, accessibility_counts.values, color=['#90EE90', '#FFD700', '#FF6347'])
ax[0].set_title('Accessibility Distribution', fontsize=14, fontweight='bold')
ax[0].set_xlabel('Accessibility Level')
ax[0].set_ylabel('Count')

sns.boxplot(x='accessibility', y='safety_index', data=df, ax=ax[1], palette='Set2')
ax[1].set_title('Safety Index by Accessibility', fontsize=14, fontweight='bold')
ax[1].set_xlabel('Accessibility')
ax[1].set_ylabel('Safety Index')

plt.tight_layout()
plt.savefig('visualizations/08_accessibility_safety.png', dpi=300, bbox_inches='tight')
print("✅ Saved: 08_accessibility_safety.png")
plt.close()

```

```python
# 9. DEVICE & ENGAGEMENT
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

device_counts = df['device_type'].value_counts()
ax[0].pie(device_counts, labels=device_counts.index, autopct='%1.1f%%', 
          startangle=90, colors=['#ff9999', '#66b3ff'])
ax[0].set_title('Device Type Distribution', fontsize=14, fontweight='bold')

sns.violinplot(x='device_type', y='engagement_score', data=df, ax=ax[1], palette='muted')
ax[1].set_title('Engagement Score by Device', fontsize=14, fontweight='bold')
ax[1].set_xlabel('Device Type')
ax[1].set_ylabel('Engagement Score')

plt.tight_layout()
plt.savefig('visualizations/09_device_engagement.png', dpi=300, bbox_inches='tight')
print("✅ Saved: 09_device_engagement.png")
plt.close()

```
