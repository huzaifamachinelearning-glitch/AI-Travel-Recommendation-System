```python
import pandas as pd
import numpy as np
import random

np.random.seed(42)
places = ['Goa', 'Manali', 'Jaipur', 'Kerala', 'Leh', 'Rishikesh', 'Udaipur', 'Darjeeling', 'Varanasi', 'Mysore']
states = ['Goa', 'Himachal Pradesh', 'Rajasthan', 'Kerala', 'Ladakh', 'Uttarakhand', 'Rajasthan', 'West Bengal', 'Uttar Pradesh', 'Karnataka']
latitudes = [15.2993, 32.2396, 26.9124, 10.8505, 34.1526, 30.0869, 24.5854, 27.0360, 25.3176, 12.2958]
longitudes = [74.1240, 77.1887, 75.7873, 76.2711, 77.5770, 78.2676, 73.7125, 88.2627, 82.9739, 76.6394]
place_types = ['Beach', 'Hill Station', 'Historical', 'Backwater', 'Adventure', 'Spiritual', 'Heritage', 'Hill Station', 'Spiritual', 'Heritage']
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
climates = ['Tropical', 'Temperate', 'Arid', 'Humid', 'Cold']
companions = ['Solo', 'Family', 'Friends', 'Couple']
interests = ['Culture', 'Adventure', 'Relaxation', 'Nature', 'Food', 'Spiritual']

def generate_dataset(n_rows=20000):
    data = []
    for i in range(n_rows):
        idx = np.random.randint(0, len(places))
        place_id = i + 1
        place = places[idx]
        state = states[idx]
        lat = latitudes[idx] + np.random.normal(0, 0.1)
        lon = longitudes[idx] + np.random.normal(0, 0.1)
        ptype = place_types[idx]
        month = random.choice(months)
        climate = random.choice(climates)
        popularity = round(np.random.uniform(1, 5), 1)
        accessibility = np.random.choice(['Easy', 'Moderate', 'Hard'])
        budget = np.random.randint(5, 50)  # Thousands INR per day
        trip_duration = np.random.randint(1, 15)
        companion = random.choice(companions)
        interest = random.choice(interests)
        safety = round(np.random.uniform(1, 5), 1)
        entry_fee = np.random.choice([0, 10, 20, 50, 100, 150])
        user_id = np.random.randint(1, 5000)
        session_id = np.random.randint(100000, 999999)
        rating_given = round(np.random.uniform(1, 5), 1)
        review_text = random.choice([
            "Amazing place!", 
            "Had a wonderful time.", 
            "Could be better.", 
            "Not worth the price.", 
            "Great for families."
        ])
        travel_start = pd.Timestamp('2025-01-01') + pd.to_timedelta(np.random.randint(0, 365), unit='d')
        travel_end = travel_start + pd.Timedelta(days=trip_duration)
        device = np.random.choice(['mobile', 'desktop'])
        engagement = np.random.randint(1, 100)
        mobility = random.choice(['None', 'Wheelchair', 'Kids'])
        language = random.choice(['English', 'Hindi', 'Other'])

        data.append([place_id, place, state, lat, lon, ptype, month, climate, popularity,
                     accessibility, budget, trip_duration, companion, interest, safety,
                     entry_fee, user_id, session_id, rating_given, review_text,
                     travel_start, travel_end, device, engagement, mobility, language])

    columns = ['place_id', 'place_name', 'state', 'latitude', 'longitude', 'place_type', 'month', 'climate', 'popularity_level',
               'accessibility', 'budget_thousand_inr', 'trip_duration_days', 'companions', 'interest', 'safety_index',
               'entry_fee', 'user_id', 'session_id', 'rating_given', 'review_text',
               'travel_dates_start', 'travel_dates_end', 'device_type', 'engagement_score', 'mobility_restrictions', 'language_preference']

    df = pd.DataFrame(data, columns=columns)
    return df

# Generate dataset
df = generate_dataset()
df.to_csv('synthetic_travel_recommendation_dataset.csv', index=False)
print("Synthetic dataset generated with shape:", df.shape)
```
